**Problem 2.** Cleaning data and dealing with categorical features (Total Marks: 40)


**Q1.** 

- Use an appropriate `pandas` function to impute missing values using one of the following two strategies: `mean` and `mode`. (10 marks)
    - Take into consideration the type of each variable  and the best practices we discussed in class/lecture notes
- Explain what data imputation is, how you have done it here, and what decisions you had to make. (5 marks)



(Total: 15 marks)

```python
# ---- provide your code here -----
from sklearn.impute import SimpleImputer
import numpy as np

numeric = ['LIMIT_BAL', 'AGE', 'BILL_AMT1', 'BILL_AMT2', 'BILL_AMT3', 'BILL_AMT4', 'BILL_AMT5', 'BILL_AMT6', 'PAY_AMT1', 'PAY_AMT2', 'PAY_AMT3', 'PAY_AMT4', 'PAY_AMT5', 'PAY_AMT6']
ordinal = ['EDUCATION', 'PAY_0', 'PAY_2', 'PAY_3', 'PAY_4', 'PAY_5', 'PAY_6']
nominal = ['SEX', 'MARRIAGE', 'default payment next month']

for column in numeric:
    df[column].fillna(df[column].mean(axis = 0), inplace=True)
    
for column in nominal + ordinal:
    df[column].fillna(df[column].mode()[0], inplace=True)
```

---- provide your text answer here ----

Data imputation is a method to deal with missing values in the dataset by replacing with estimated or calculated values. We can use different interpolation to fill in the missing observations, namely mean, median and mode. 

Here, I have to decide which interpolation I should use for different variables. 
- For numeric variables, I used **mean** since the distribution is symmetrical and no significant effect by outliers. 
- For the categorical variables like ordinal and nominal, I used **mode** as it gives the most frequent value, which is a better estimator of each feature. 

By replacing missing values with mean for numaric and mode for categorical features using `for`loop and `fillna` method, the data is now complete with no gaps in the data.

**Problem 3** Preparing X and y arrays (Total Marks: 10)

**Q1**. 

- Create a numpy array `y` from the first 7,500 observations of 'payment_default' column from `df` (2.5 marks)   
- Create a numpy array `X`  from the first 7,500 observations of all the remaining variables in `df` (2.5 marks)   

(Total: 5 Marks)

```python
# ---- provide your code here -----
y = df.loc[0:7499, 'default payment next month'].values
X = df.loc[0:7499, df.columns != 'default payment next month'].values
```

<hr style="width:25%;margin-left:0;"> 

**Q2**. 

- Use an appropriate `sklearn` library we used in class to create `y_train`, `y_test`, `X_train` and `X_test` by splitting the data into 75% train and 25% test datasets (2.5 marks) 
    - Set random_state to 3 and stratify the subsamples so that train and test datasets have roughly equal proportions of the target's class labels 
- Standardise the data to mean zero and variance one using an approapriate `sklearn` library (2.5 marks)   

(Total: 5 marks) 

```python
# ---- provide your code here -----
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler

X_train, X_test, y_train, y_test = train_test_split(X, y, test_size = 0.25, random_state = 3, stratify = y)

sc = StandardScaler()
sc.fit(X_train)

X_train_scaled = sc.fit_transform(X_train)
X_test_scaled = sc.transform(X_test)
```

---
---

**Problem 4**. Support Vector Classifier and Accuracies (Total Marks: 30)


**Q1**. 

- Train a Support Vector Classifier on the standardised data (5 marks)
    - Use `rbf` kernel and set `random_state` to 24 (don't change any other parameters)
- Compute and print training and test dataset accuracies (5 marks)

(Total: 10 marks)   

```python
# ---- provide your code here -----

from sklearn.svm import SVC
from sklearn.metrics import accuracy_score

svm_clf = SVC(kernel='rbf', random_state=24)
svm_clf.fit(X_train_scaled, y_train)

# Compute training accuracy
train_preds = svm_clf.predict(X_train_scaled)
train_accuracy = accuracy_score(y_train, train_preds)

# Compute test accuracy
test_preds = svm_clf.predict(X_test_scaled)
test_accuracy = accuracy_score(y_test, test_preds)

# Print training and test dataset accuracies
print(f'SupportVector Train Accuracy: {train_accuracy:.4f}')
print(f'SupportVector Test Accuracy: {test_accuracy:.4f}')
```

<hr style="width:25%;margin-left:0;"> 

**Q2.**

- Extract 2 linear principal components from the standardised features using an appropriate `sklearn` library (5 marks)
- Train a Support Vector Classifier on the 2 principal components computed above (5 marks)   
    - Use `rbf` kernel and set `random_state` to 24 (don't change any other parameters)
- Compute and print training and test dataset accuracies (5 marks)


(Total: 15 marks)  

```python
# ---- provide your code here -----
from sklearn.decomposition import PCA
from sklearn.svm import SVC

# Extract 2 linear principal components
pca = PCA(n_components=2)
X_train_pca = pca.fit_transform(X_train_scaled)
X_test_pca = pca.transform(X_test_scaled)

# Train Support Vector Classifier on the 2 principal components with RBF kernel
svm_clf_pca = SVC(kernel='rbf', random_state=24)
svm_clf_pca.fit(X_train_pca, y_train)

# Compute training accuracy
train_preds_pca = svm_clf_pca.predict(X_train_pca)
train_accuracy_pca = accuracy_score(y_train, train_preds_pca)

# Compute test accuracy
test_preds_pca = svm_clf_pca.predict(X_test_pca)
test_accuracy_pca = accuracy_score(y_test, test_preds_pca)

# Print training and test dataset accuracies
print(f'SupportVector Train Accuracy (PCA): {train_accuracy_pca:.4f}')
print(f'SupportVector Test Accuracy (PCA): {test_accuracy_pca:.4f}')
```
